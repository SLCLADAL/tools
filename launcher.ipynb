# LADAL Tools Launcher

**Welcome to the LADAL interactive tools environment.**

To launch a tool:

1. Find the tool you want below
2. Run its code cell (click the cell, then press **Shift+Enter**)
3. Wait for the **✔ Ready!** message
4. Click the link that appears to open the tool

---

## 🔍 WordFinder
**Keyword-in-context concordancing** — upload texts and search for words and phrases

```python
import subprocess, time, os
from IPython.display import display, HTML

port = 3838
path = '/home/jovyan/tools/wordfinder'

print('Starting WordFinder... (this takes about 20 seconds)')
subprocess.Popen(['R', '-e', f"shiny::runApp('{path}', port={port}, host='0.0.0.0')"],
                 stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

# Wait for the app to be ready
import socket
for i in range(60):
    try:
        s = socket.create_connection(('localhost', port), timeout=1)
        s.close()
        break
    except:
        time.sleep(1)
        print(f'  waiting... ({i+1}s)', end='\r')

# Build the proxy URL from the current notebook URL
base_url = os.environ.get('JUPYTERHUB_SERVICE_PREFIX', '/')
tool_url = f'{base_url}proxy/{port}/'

display(HTML(f'''
<div style="margin:16px 0; padding:16px; background:#eafaf1;
            border-left:4px solid #27ae60; border-radius:6px;">
  <p style="margin:0; font-family:sans-serif; font-size:1rem;">
    ✔ <b>WordFinder is ready!</b><br><br>
    <a href="{tool_url}" target="_blank"
       style="display:inline-block; padding:10px 24px;
              background:#51247a; color:white;
              border-radius:7px; text-decoration:none;
              font-weight:bold; font-size:1rem;">
      🔍 Open WordFinder
    </a>
  </p>
</div>
'''))
```
